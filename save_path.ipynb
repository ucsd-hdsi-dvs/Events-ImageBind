```python
import os
import pickle
import random
from sklearn.model_selection import train_test_split

# Configuration
base_dir = "/eastdata/datasets/sd-data/fler-guide"
subfolders = ['fler', 'gray0', 'gray1']
random_seed = 42  # For reproducibility
test_size = 0.2   # 20% validation, 80% training

# Set random seed for reproducibility
random.seed(random_seed)

# Step 1: Get sorted filenames from 'fler'
fler_dir = os.path.join(base_dir, 'fler')
if not os.path.exists(fler_dir):
    raise FileNotFoundError(f"Folder not found: {fler_dir}")

# Sort files (ignoring hidden files and directories)
fler_files = sorted([
    f for f in os.listdir(fler_dir) 
    if not f.startswith('.') and os.path.isfile(os.path.join(fler_dir, f))
])

# Step 2: Split into train/validation sets
train_files, val_files = train_test_split(fler_files, test_size=test_size, random_state=random_seed)

# Step 3: Build path dictionaries for train and validation
def build_path_dict(file_list):
    path_dict = {}
    for subfolder in subfolders:
        folder_path = os.path.join(base_dir, subfolder)
        path_dict[subfolder] = [os.path.join(folder_path, f) for f in file_list]
        
        # Verify files exist
        for path in path_dict[subfolder]:
            if not os.path.exists(path):
                raise FileNotFoundError(f"File missing: {path}")
    return path_dict

train_paths = build_path_dict(train_files)
val_paths = build_path_dict(val_files)

# Step 4: Save to pickle files
with open("/eastdata/datasets/sd-data/fler-guide/train_paths.pkl", 'wb') as f:
    pickle.dump(train_paths, f)

with open("/eastdata/datasets/sd-data/fler-guide/val_paths.pkl", 'wb') as f:
    pickle.dump(val_paths, f)

print(f"Training set: {len(train_files)} files")
print(f"Validation set: {len(val_files)} files")
print(f"Saved to train_paths.pkl and val_paths.pkl")
```
